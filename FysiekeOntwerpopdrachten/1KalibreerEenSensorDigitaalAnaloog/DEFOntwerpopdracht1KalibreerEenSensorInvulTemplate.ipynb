## *Opdracht 6*: Meten en fitten, Leerdoel

```python
import numpy as np
array_temps = np.array([521, 510, 524, 728, 729, 732, 407, 422, 404, 690, 660, 680, 190, 182, 260])
array_control_temps = np.array([23.4, 23.6, 23.6, 4.5, 4.3, 4.1, 33.2, 31.5, 33.8, 4.7, 5.8, 5.2, 63.5, 65.3, 51.6])

x = np.array([499, 497, 508, 509, 504, 510, 488, 495, 506, 504]) #Metingen gedaan in zelfde omgeving (kamertemperatuur)
```

```python
from scipy.optimize import curve_fit
import matplotlib.pyplot as plt

def linear_model(x, a, b):
    return a*x + b

popt,pcov = curve_fit(linear_model,array_temps,array_control_temps)
a=popt[0]
b=popt[1]

x_fit = np.linspace(100,900,1000)
y_fit = a*x_fit+b

sigma_x = np.std(x)
N = len(x)

x_error = sigma_x / np.sqrt(N)
y_error = 0.1 #Deviatie voor thermometer controletemperatuur, gevonden op internet

plt.figure()
plt.plot(x_fit,y_fit)
plt.plot(array_temps,array_control_temps, 'k.', label='gemeten data')
plt.errorbar(array_temps,array_control_temps, xerr=x_error, yerr=y_error, fmt='y.', label="fitfunctie")
plt.legend()
plt.xlim(100,900)
plt.ylim(-5,75)
plt.xlabel('Bits')
plt.ylabel('Controletemperatuur')
plt.show
```
