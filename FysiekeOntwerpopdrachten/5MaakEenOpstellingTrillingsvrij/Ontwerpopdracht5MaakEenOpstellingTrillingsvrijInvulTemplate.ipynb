```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.
g = 9.81
w0 = 2*np.pi * f0
deltax = g/w0**2
print("Delta-x is",deltax,"m")
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

m = np.linspace(0,20,1000)
C = m*g/deltax

deltax2 = 2 * deltax
C2 = m*g/deltax2

deltax3 = 5 * deltax
C3 = m*g/deltax3
plt.figure()
plt.plot(m,C,"k.",label="1*delta x")
plt.plot(m,C2,"y.",label="2*delta x")
plt.plot(m,C3,"b.",label="5*delta x")
plt.xlabel("m (kg)")
plt.ylabel("C (N/m)")
plt.legend()
plt.show()
plt.savefig("grafiek")
```

## Opdracht 7: Plotten maar

```python
# Voeg hier je plots in.

f0 = 2.5
f = np.linspace(0,5,1000)


C_real = 25 #N/m
m_real = 0.5 #kg
w0_real = np.sqrt(C_real/m_real)
f0_real = w0_real/(2*np.pi)
Am_Av_real = np.abs(1/(1-(f**2/f0_real**2)))
Am_arr = np.array([0.76,0.86,0.96,1.15])
Av_arr = np.array([5,5.89,1.29,5.1])
Am_Av_arr = Am_arr/Av_arr
f_arr = np.array([5.3,3,1.4,2.7])
print(f0_real)
plt.figure()
plt.loglog(f,Am_Av_real,'r-',label='real')
plt.errorbar(f_arr,Am_Av_arr,xerr=0.2,yerr=0.4,fmt='k.',label='metingen')
plt.axvline(f0_real)
plt.xlabel("log(f) (log(Hz))")
plt.ylabel("log(Am/Av)")
plt.legend()
plt.show()
```
